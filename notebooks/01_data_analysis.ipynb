```python

import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader

print("PyTorch version:", torch.__version__)


device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Training device:", device)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
```

```python
mlp_features = [
    "Product_ID",
    "Category_ID",
    "Lag_1",
    "Lag_2",
    "Lag_3",
    "Lag_4",
    "Rolling_Mean_4",
    "Rolling_Std_4",
    "Stock_Quantity",
    "Reorder_Level",
    "Reorder_Quantity",
    "Unit_Price"
]

target = "Target_Next_Week"

X_train = train_df[mlp_features].values
y_train = train_df[target].values

X_val = val_df[mlp_features].values
y_val = val_df[target].values

X_test = test_df[mlp_features].values
y_test = test_df[target].values

print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("X_val:", X_val.shape)
print("y_val:", y_val.shape)

print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
```

```python
X_train_tensor = torch.tensor(
    X_train,
    dtype=torch.float32
)

y_train_tensor = torch.tensor(
    y_train,
    dtype=torch.float32
).view(-1, 1)

X_val_tensor = torch.tensor(
    X_val,
    dtype=torch.float32
)

y_val_tensor = torch.tensor(
    y_val,
    dtype=torch.float32
).view(-1, 1)

X_test_tensor = torch.tensor(
    X_test,
    dtype=torch.float32
)

y_test_tensor = torch.tensor(
    y_test,
    dtype=torch.float32
).view(-1, 1)

print("Tensors created successfully!")
```

```python


batch_size = 16

train_dataset = TensorDataset(
    X_train_tensor,
    y_train_tensor
)

val_dataset = TensorDataset(
    X_val_tensor,
    y_val_tensor
)

test_dataset = TensorDataset(
    X_test_tensor,
    y_test_tensor
)

train_loader = DataLoader(
    train_dataset,
    batch_size=batch_size,
    shuffle=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=batch_size,
    shuffle=False
)

test_loader = DataLoader(
    test_dataset,
    batch_size=batch_size,
    shuffle=False
)

print("DataLoaders created successfully!")
print("Batch size:", batch_size)
```

```python
class DemandMLP(nn.Module):

    def __init__(self, input_size):
        super().__init__()

        self.network = nn.Sequential(
            nn.Linear(input_size, 64),
            nn.ReLU(),

            nn.Dropout(0.2),

            nn.Linear(64, 32),
            nn.ReLU(),

            nn.Linear(32, 16),
            nn.ReLU(),

            nn.Linear(16, 1)
        )

    def forward(self, x):
        return self.network(x)
mlp_model = DemandMLP(
    input_size=len(mlp_features)
).to(device)

print(mlp_model)
```

```python
criterion = nn.MSELoss()

optimizer = torch.optim.Adam(
    mlp_model.parameters(),
    lr=0.001,
    weight_decay=1e-5
)

print("Loss function:", criterion)
print("Optimizer:", optimizer)
```

```python

num_epochs = 100

train_losses = []
val_losses = []

for epoch in range(num_epochs):

    
    mlp_model.train()

    running_train_loss = 0.0

    for X_batch, y_batch in train_loader:

        X_batch = X_batch.to(device)
        y_batch = y_batch.to(device)

      
        optimizer.zero_grad()

       
        predictions = mlp_model(X_batch)

       
        loss = criterion(predictions, y_batch)

        
        loss.backward()

       
        optimizer.step()

        running_train_loss += loss.item() * X_batch.size(0)

    epoch_train_loss = (
        running_train_loss / len(train_loader.dataset)
    )

   
    mlp_model.eval()

    running_val_loss = 0.0

    with torch.no_grad():

        for X_batch, y_batch in val_loader:

            X_batch = X_batch.to(device)
            y_batch = y_batch.to(device)

            predictions = mlp_model(X_batch)

            loss = criterion(predictions, y_batch)

            running_val_loss += (
                loss.item() * X_batch.size(0)
            )

    epoch_val_loss = (
        running_val_loss / len(val_loader.dataset)
    )

    train_losses.append(epoch_train_loss)
    val_losses.append(epoch_val_loss)

  
    if (epoch + 1) % 10 == 0:
        print(
            f"Epoch [{epoch+1}/{num_epochs}] "
            f"Train Loss: {epoch_train_loss:.4f} "
            f"Val Loss: {epoch_val_loss:.4f}"
        )
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    train_losses,
    label="Training Loss"
)

plt.plot(
    val_losses,
    label="Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("MSE Loss")
plt.title("PyTorch MLP Training Progress")

plt.legend()
plt.tight_layout()
plt.show()
```

```python
mlp_model.eval()
val_predictions = []

with torch.no_grad():

    for X_batch, _ in val_loader:

        X_batch = X_batch.to(device)

        predictions = mlp_model(X_batch)

        val_predictions.extend(
            predictions.cpu().numpy().flatten()
        )

val_predictions = np.array(val_predictions)

print("Validation predictions:", val_predictions.shape)
print("Actual validation values:", y_val.shape)
```

```python


val_mae = mean_absolute_error(
    y_val,
    val_predictions
)

val_rmse = np.sqrt(
    mean_squared_error(
        y_val,
        val_predictions
    )
)

val_wape = (
    np.sum(np.abs(y_val - val_predictions))
    /
    np.sum(np.abs(y_val))
) * 100

val_smape = (
    np.mean(
        2 * np.abs(y_val - val_predictions)
        /
        (
            np.abs(y_val)
            + np.abs(val_predictions)
            + 1e-8
        )
    )
) * 100

print("MLP Validation Results")
print("----------------------")
print("MAE  :", round(val_mae, 2))
print("RMSE :", round(val_rmse, 2))
print("WAPE :", round(val_wape, 2), "%")
print("sMAPE:", round(val_smape, 2), "%")
```

```python
lstm_df = product_weekly.copy()
lstm_df = lstm_df.sort_values(
    ["Product_Name", "Week"]
).reset_index(drop=True)


lstm_df["Gap_Days"] = (
    lstm_df.groupby("Product_Name")["Week"]
    .diff()
    .dt.days
)


lstm_df["Gap_Days"] = (
    lstm_df["Gap_Days"]
    .fillna(0)
)

print("LSTM source shape:", lstm_df.shape)

lstm_df.head(10)
```

```python
class DemandLSTM(nn.Module):

    def __init__(
        self,
        input_size=2,
        hidden_size=64,
        num_layers=2,
        dropout=0.2
    ):
        super().__init__()

        self.lstm = nn.LSTM(
            input_size=input_size,
            hidden_size=hidden_size,
            num_layers=num_layers,
            batch_first=True,
            dropout=dropout
        )

        self.fc = nn.Sequential(
            nn.Linear(hidden_size, 32),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(32, 1)
        )

    def forward(self, x):

        lstm_output, _ = self.lstm(x)
        last_output = lstm_output[:, -1, :]
        output = self.fc(last_output)
        return output

lstm_model = DemandLSTM(
    input_size=2,
    hidden_size=64,
    num_layers=2,
    dropout=0.2
).to(device)

print(lstm_model)
```

```python
# ==========================================
# STEP 58: ADAPTIVE/HYBRID PYTORCH MODEL
# ==========================================

class AdaptiveHybridDemandModel(nn.Module):

    def __init__(
        self,
        num_products,
        sequence_input_size=6,
        embedding_dim=16,
        hidden_size=64
    ):
        super().__init__()

        # ----------------------------------
        # Product embedding
        # ----------------------------------
        self.product_embedding = nn.Embedding(
            num_embeddings=num_products,
            embedding_dim=embedding_dim
        )

        # ----------------------------------
        # Sequential demand/inventory branch
        # ----------------------------------
        self.lstm = nn.LSTM(
            input_size=sequence_input_size,
            hidden_size=hidden_size,
            num_layers=2,
            batch_first=True,
            dropout=0.2
        )

        # ----------------------------------
        # Product embedding encoder
        # ----------------------------------
        self.product_encoder = nn.Sequential(
            nn.Linear(embedding_dim, 16),
            nn.ReLU()
        )

        # ----------------------------------
        # Fusion + prediction
        # ----------------------------------
        self.fusion = nn.Sequential(
            nn.Linear(
                hidden_size + 16,
                64
            ),
            nn.ReLU(),

            nn.Dropout(0.2),

            nn.Linear(64, 32),
            nn.ReLU(),

            nn.Linear(32, 1)
        )

    def forward(self, x_sequence, product_id):

        # LSTM branch
        lstm_output, _ = self.lstm(
            x_sequence
        )

        # Final timestep
        sequence_representation = (
            lstm_output[:, -1, :]
        )

        # Product embedding branch
        product_embedding = (
            self.product_embedding(product_id)
        )

        product_representation = (
            self.product_encoder(
                product_embedding
            )
        )

        # Combine both representations
        combined = torch.cat(
            [
                sequence_representation,
                product_representation
            ],
            dim=1
        )

        # Final prediction
        output = self.fusion(combined)

        return output


# Create model
hybrid_model = AdaptiveHybridDemandModel(
    num_products=num_products,
    sequence_input_size=6,
    embedding_dim=16,
    hidden_size=64
).to(device)

print(hybrid_model)
```

```python
# ==========================================
# STEP 60: HYBRID LOSS & OPTIMIZER
# ==========================================

hybrid_criterion = nn.MSELoss()

hybrid_optimizer = torch.optim.Adam(
    hybrid_model.parameters(),
    lr=0.001,
    weight_decay=1e-5
)

print("Hybrid loss:", hybrid_criterion)
print("Hybrid optimizer:", hybrid_optimizer)
```

```python
# ==========================================
# STEP 61: TRAIN HYBRID MODEL
# ==========================================

num_epochs_hybrid = 100

hybrid_train_losses = []
hybrid_val_losses = []

best_val_loss = float("inf")
best_model_state = None

patience = 15
patience_counter = 0

for epoch in range(num_epochs_hybrid):

    # ----------------------------------
    # Training
    # ----------------------------------
    hybrid_model.train()

    running_train_loss = 0.0

    for X_batch, product_batch, y_batch in hybrid_train_loader:

        X_batch = X_batch.to(device)
        product_batch = product_batch.to(device)
        y_batch = y_batch.to(device)

        hybrid_optimizer.zero_grad()

        predictions = hybrid_model(
            X_batch,
            product_batch
        )

        loss = hybrid_criterion(
            predictions,
            y_batch
        )

        loss.backward()

        # Prevent exploding gradients
        torch.nn.utils.clip_grad_norm_(
            hybrid_model.parameters(),
            max_norm=1.0
        )

        hybrid_optimizer.step()

        running_train_loss += (
            loss.item() * X_batch.size(0)
        )

    epoch_train_loss = (
        running_train_loss
        / len(hybrid_train_loader.dataset)
    )

    # ----------------------------------
    # Validation
    # ----------------------------------
    hybrid_model.eval()

    running_val_loss = 0.0

    with torch.no_grad():

        for X_batch, product_batch, y_batch in hybrid_val_loader:

            X_batch = X_batch.to(device)
            product_batch = product_batch.to(device)
            y_batch = y_batch.to(device)

            predictions = hybrid_model(
                X_batch,
                product_batch
            )

            loss = hybrid_criterion(
                predictions,
                y_batch
            )

            running_val_loss += (
                loss.item() * X_batch.size(0)
            )

    epoch_val_loss = (
        running_val_loss
        / len(hybrid_val_loader.dataset)
    )

    hybrid_train_losses.append(
        epoch_train_loss
    )

    hybrid_val_losses.append(
        epoch_val_loss
    )

    # ----------------------------------
    # Save best model
    # ----------------------------------
    if epoch_val_loss < best_val_loss:

        best_val_loss = epoch_val_loss

        best_model_state = {
            key: value.detach().cpu().clone()
            for key, value
            in hybrid_model.state_dict().items()
        }

        patience_counter = 0

    else:

        patience_counter += 1

    # ----------------------------------
    # Progress
    # ----------------------------------
    if (epoch + 1) % 10 == 0:

        print(
            f"Epoch [{epoch+1}/{num_epochs_hybrid}] "
            f"Train Loss: {epoch_train_loss:.4f} "
            f"Val Loss: {epoch_val_loss:.4f}"
        )

    # ----------------------------------
    # Early stopping
    # ----------------------------------
    if patience_counter >= patience:

        print(
            f"\nEarly stopping at epoch "
            f"{epoch + 1}"
        )

        break


# Restore best validation model
if best_model_state is not None:

    hybrid_model.load_state_dict(
        best_model_state
    )

    hybrid_model.to(device)

print("\nBest validation loss:",
      round(best_val_loss, 4))
```

```python
# ==========================================
# STEP 76.2: SAVE DASHBOARD DATASET
# ==========================================

dashboard_path = (
    "../results/smartstock_dashboard_data.csv"
)

dashboard_df.to_csv(
    dashboard_path,
    index=False
)

print("Dashboard dataset saved successfully!")
print("Location:", dashboard_path)
```
